### Replace  all invalid  data  with median 

```python
df.loc[df['Shipping_Days'] < 0, 'Shipping_Days'] = df['Shipping_Days'].median()

df.loc[df['Customer_Rating'] > 5, 'Customer_Rating'] = df['Customer_Rating'].median()

df.loc[df['Quantity'] <= 0, 'Quantity'] = df['Quantity'].median()

df.loc[df['Estimated_Cost'] < 0, 'Estimated_Cost'] = df['Estimated_Cost'].median()

df.loc[df['Net_Sales'] < 0, 'Net_Sales'] = df['Net_Sales'].median()

df.loc[df['Discount_Amount'] < 0, 'Discount_Amount'] = df['Discount_Amount'].median()

df.loc[df['Gross_Amount'] < 0, 'Gross_Amount'] = df['Gross_Amount'].median()

df.loc[(df['Discount_Pct'] < 0) | (df['Discount_Pct'] > 100), 'Discount_Pct'] = df['Discount_Pct'].median()
```

```python
import matplotlib.pyplot as plt
```

```python
highest_product = df.groupby('Product_Name')['Quantity'].sum().sort_values(ascending=False)

pd = highest_product.plot(
    kind='bar',
    figsize=(10, 5),
    color=['red', 'blue', 'green', 'orange', 'purple',
           'pink', 'brown', 'cyan', 'yellow', 'gray']
)
pd.bar_label(pd.containers[0])
plt.xlabel('Product Name')
plt.ylabel('Total Quantity Sold')
plt.title('Top 10 Selling Products')
plt.xticks(rotation=45)
plt.show()
```

```python
profit = df.groupby('Product_Name')['Profit'].sum().sort_values(ascending=False).head(10)

ax = profit.plot(
    kind='bar',
    figsize=(10, 5),
    color=['red', 'blue', 'green', 'orange', 'purple',
           'pink', 'brown', 'cyan', 'yellow', 'gray']
)

ax.bar_label(ax.containers[0])

plt.xlabel('Product Name')
plt.ylabel('Total Profit')
plt.title('Top 10 Products by Profit')
plt.xticks(rotation=45)
plt.show()
```

```python
Quantity_profit= df.groupby('Product_Name').agg(
    Quantity=('Quantity', 'sum'),
    Profit=('Profit', 'sum')
).sort_values('Profit', ascending=False)

colors = ['red', 'blue', 'green', 'orange', 'purple',
          'pink', 'brown', 'cyan', 'gold', 'gray']

plt.figure(figsize=(10, 6))

for i, product in enumerate(Quantity_profit.index):
    plt.scatter(
        Quantity_profit.loc[product, 'Quantity'],
        Quantity_profit.loc[product, 'Profit'],
        color=colors[i],
        label=product
    )

plt.xlabel('Total Quantity Sold')
plt.ylabel('Total Profit')
plt.title('Quantity vs Profit by Product')
plt.grid(True)
plt.legend()
plt.show()
```

```python
payment_counts = df['Payment_Method'].dropna().value_counts()

plt.figure(figsize=(8, 8))

plt.pie(
    payment_counts.values,
    labels=payment_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Payment Method Distribution')
plt.legend(title='Payment Method')
plt.show()
```

```python
marketing_sales = df.groupby('Marketing_Source')['Net_Sales'].sum()
plt.figure(figsize=(10, 7))
plt.plot(
    marketing_sales.index,
    marketing_sales.values,
    marker='o'
)

for source, sales in marketing_sales.items():
    plt.annotate(
        f'{sales:,.0f}',
        (source, sales),
        textcoords="offset points",
        xytext=(0,8),
        ha='center'
    )

plt.xlabel('Marketing Source')
plt.ylabel('Total Sales')
plt.title('Marketing Source vs Total Sales')
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

financial_cols = [
    'Unit_Price',
    'Gross_Amount',
    'Discount_Amount',
    'Net_Sales',
    'Estimated_Cost',
    'Profit',
    'Profit_Margin_Pct'
]

corr = df[financial_cols].corr()

plt.figure(figsize=(10, 7))

sns.heatmap(
    corr,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Financial Columns')
plt.show()
```
