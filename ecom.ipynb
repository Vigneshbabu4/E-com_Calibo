### Replace  all invalid  data  with median

```python
df.loc[df['Shipping_Days'] < 0, 'Shipping_Days'] = df['Shipping_Days'].median()

df.loc[df['Customer_Rating'] > 5, 'Customer_Rating'] = df['Customer_Rating'].median()

df.loc[df['Quantity'] <= 0, 'Quantity'] = df['Quantity'].median()

df.loc[df['Estimated_Cost'] < 0, 'Estimated_Cost'] = df['Estimated_Cost'].median()

df.loc[df['Net_Sales'] < 0, 'Net_Sales'] = df['Net_Sales'].median()

df.loc[df['Discount_Amount'] < 0, 'Discount_Amount'] = df['Discount_Amount'].median()

df.loc[df['Gross_Amount'] < 0, 'Gross_Amount'] = df['Gross_Amount'].median()

df.loc[(df['Discount_Pct'] < 0) | (df['Discount_Pct'] > 100), 'Discount_Pct'] = df['Discount_Pct'].median()
```

```python
top_10 = df.groupby('Product_Name').agg(
    Quantity=('Quantity', 'sum'),
    Profit=('Profit', 'sum')
).sort_values('Profit', ascending=False)

colors = ['red', 'blue', 'green', 'orange', 'purple',
          'pink', 'brown', 'cyan', 'gold', 'gray']

plt.figure(figsize=(10, 6))

for i, product in enumerate(top_10.index):
    plt.scatter(
        top_10.loc[product, 'Quantity'],
        top_10.loc[product, 'Profit'],
        color=colors[i],
        label=product
    )

plt.xlabel('Total Quantity Sold')
plt.ylabel('Total Profit')
plt.title('Quantity vs Profit by Product')
plt.grid(True)
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

return_category = pd.crosstab(
    df["Category"],
    df["Return_Status"]
)

RC = return_category.plot(
    kind="bar",
    figsize=(10,6),
    color=["green", "red"]
)

for container in RC.containers:
    RC.bar_label(container, padding=3)

plt.xlabel("Category")
plt.ylabel("Number of Orders")
plt.title("Returns by Category")
plt.xticks(rotation=45)
plt.show()
```

```python
product_profit = df.groupby('Product_Name')['Profit'].sum().sort_values()
print(product_profit)
```

```python
def recommendation(profit):
    if profit < product_profit.median():
        return 'Low Profit - Consider Alternative'
    else:
        return 'Good Profit - Recommended'

result = product_profit.reset_index()
result['Recommendation'] = result['Profit'].apply(recommendation)

print(result)
```

```python
profit = df.groupby('Product_Name')['Profit'].sum().sort_values()

plt.figure(figsize=(10, 6))

bars = plt.bar(profit.index, profit.values)

plt.bar_label(bars, fmt='%.0f')

plt.xlabel('Product Name')
plt.ylabel('Total Profit')
plt.title('Product Profitability and Recommendation')
plt.xticks(rotation=45)
plt.grid(axis='y')

plt.show()
```
